```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

##### Penjelasan:
1. import numpy as np -> mengimpor NumPy untuk melakukan pengolahan data dalam bnetuk array
2. import pandas as pd -> Mengimpor Pandas untuk menolah dan menampilkan data dalam bentuk tabel
3. from sklearn.datasets import load_iris -> Mengimpor fungsi load_iris dari scikit-learn untuk mengambil dataset iris
4. iris = load_iris() → Memuat dataset Iris dan menyimpannya ke dalam variabel iris agar dapat digunakan untuk proses klasifikasi selanjutnya.

```python
#konversi data menjadi dataframe pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

#defenisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

##### Penjelasan
1. pd.DataFrame(...) → Mengubah data Iris menjadi DataFrame dan menggunakan nama fitur sebagai nama kolom.
2. df["Species"] → Menambahkan kolom Species yang berisi jenis bunga Iris.
3. X = df.drop(...) → Menentukan X sebagai data fitur dengan menghapus kolom Species.
4. y = df["Species"] → Menentukan y sebagai target atau label berupa jenis bunga.
5. print(X) → Menampilkan data fitur yang digunakan dalam proses klasifikasi.

```python
y
```

##### Penjelasan
y : menampilkan isi label. Terlihat masih berupa teks (setosa, virginica, dst.) dengan tipe data object.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

##### Penjelasan
1. lab = LabelEncoder() : membuat objek encoder.
2. y = lab.fit_transform(y) : mempelajari kategori lalu mengubahnya menjadi angka:
setosa → 0
versicolor → 1
virginica → 2
3. Output berupa array berisi 0, 1, 2 (masing-masing 50 data).

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

##### Penjelasan
1. test_size=0.2 : 20% data (30 data) untuk testing, 80% (120 data) untuk training.
2. random_state=0 : mengunci pengacakan agar pembagian selalu sama setiap dijalankan (hasil bisa direproduksi).

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

##### Penjelasan
1. from sklearn.naive_bayes import GaussianNB : mengimpor algoritma Gaussian Naïve Bayes.
2. model = GaussianNB() : membuat model kosong.
3. model.fit(X_train, y_train) : melatih model. Ia menghitung rata-rata (μ) dan standar deviasi (σ) tiap fitur untuk tiap kelas, serta probabilitas awal kelas P(C).
4. NB_model : menyimpan model yang sudah terlatih.

```python
result = NB_model.predict(X_test)
result
```

##### Penjelasan
NB_model.predict(X_test) : model memprediksi kelas untuk 30 data testing.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

##### Penjelasan
1. accuracy_score(y_test, result) : membandingkan label asli (y_test) dengan hasil prediksi (result).
2. *100 : mengubah ke persen.
3. Hasil: 96,67%, artinya 29 dari 30 data testing diprediksi benar.
